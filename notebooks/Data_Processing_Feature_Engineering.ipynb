### 📐 Theoretical Foundation — Vshale and Lithology Classification

#### What is Vshale (Volume of Shale)?

**Definition:** The volumetric fraction of shale (clay minerals) within a rock formation, expressed as a decimal or percentage (0 = pure sand, 1 = pure shale).

**Petrophysical Significance:**
- **Shale:** Contains clays with high natural radioactivity (gamma ray) and low porosity
- **Sand:** Typically quartz or carbonate with low natural radioactivity and higher porosity
- **Vshale estimation:** Quantifies clay content → controls permeability, water saturation calculations, and completion strategy

**Why Matters in Formation Evaluation:**
$$\text{Permeability} \propto f(\text{Porosity}, \text{Vshale}, \text{Rock Type})$$
$$\text{Oil Saturation} \propto f(\text{Resistivity}, \text{Porosity}, \text{Vshale})$$

Incorrect Vshale → incorrect reservoir potential assessment → poor drilling/completion decisions

#### Linear GR Index Formula

The **Linear GR Index (IGR)** normalizes gamma ray measurements to a 0–1 scale:

$$I_{GR} = \frac{GR_{\text{log}} - GR_{\text{min}}}{GR_{\text{max}} - GR_{\text{min}}}$$

**Parameters:**
- $GR_{\text{log}}$ = Measured gamma ray at depth (API units)
- $GR_{\text{min}}$ = GR value of cleanest sand in well (baseline)
- $GR_{\text{max}}$ = GR value of purest shale in well (maximum)
- $I_{GR}$ = Normalized index (0 = clean sand, 1 = pure shale)

**Limitations:** Assumes linear relationship between radioactivity and clay content. Tends to **overestimate** Vshale, especially in high-clay formations.

#### Larionov Corrections (Non-Linear Models)

Larionov (1969) proposed **exponential corrections** to account for non-linearity in gamma ray response:

**For Tertiary (younger, less compacted) rocks:**
$$V_{\text{shale}} = 0.083 \times (2^{3.7 \times I_{GR}} - 1)$$

**For Pre-Tertiary (older, more compacted) rocks:**
$$V_{\text{shale}} = 0.33 \times (2^{2 \times I_{GR}} - 1)$$

**Advantages:**
- Accounts for non-linear GR response to increasing clay content
- More realistic Vshale estimates (< linear method, especially at high clay)
- Widely adopted by industry (validated through core analysis)

**Typical Vshale Thresholds for Lithology Classification:**
- $V_{\text{shale}}$ < 0.30 → **Clean Sand** (reservoir quality)
- 0.30 ≤ $V_{\text{shale}}$ < 0.50 → **Silty Sand** (mixed)
- 0.50 ≤ $V_{\text{shale}}$ < 0.80 → **Sandy Shale** (transitional)
- $V_{\text{shale}}$ ≥ 0.80 → **Shale** (seal/non-reservoir)

## 🧮 Section 4 — Calculate Vshale (Three Methods)

```python
# Define GRmin and GRmax
# In practice, we use the 5th and 95th percentile to avoid outliers
# rather than the absolute min and max
GR_min = df_clean['GR'].quantile(0.05)  # 5th percentile = clean sand
GR_max = df_clean['GR'].quantile(0.95)  # 95th percentile = pure shale

print(f'📌 Why percentiles instead of absolute min/max?')
print(f'   Absolute min/max are sensitive to outliers (bad hole, tool errors)')
print(f'   Percentiles give a more stable, robust estimate')
print(f'\n   GRmin (5th percentile)  : {GR_min:.2f} GAPI')
print(f'   GRmax (95th percentile) : {GR_max:.2f} GAPI')
print(f'   Absolute GR min         : {df_clean["GR"].min():.2f} GAPI')
print(f'   Absolute GR max         : {df_clean["GR"].max():.2f} GAPI')
```

```python
# ============================================================
# METHOD 1: Linear GR Index (IGR)
# Formula: IGR = (GR - GRmin) / (GRmax - GRmin)
# ============================================================
df_clean['IGR'] = (df_clean['GR'] - GR_min) / (GR_max - GR_min)

# Clip to 0-1 range (values outside GRmin/GRmax get clipped)
df_clean['IGR'] = df_clean['IGR'].clip(0, 1)

print('✅ Method 1: Linear IGR calculated')
print(f'   IGR range: {df_clean["IGR"].min():.4f} to {df_clean["IGR"].max():.4f}')

# ============================================================
# METHOD 2: Larionov — Tertiary (younger rocks)
# Formula: Vsh = 0.083 * (2^(3.7 * IGR) - 1)
# ============================================================
df_clean['Vsh_Tertiary'] = 0.083 * (2 ** (3.7 * df_clean['IGR']) - 1)
df_clean['Vsh_Tertiary'] = df_clean['Vsh_Tertiary'].clip(0, 1)

print('\n✅ Method 2: Larionov Tertiary calculated')
print(f'   Vsh_Tertiary range: {df_clean["Vsh_Tertiary"].min():.4f} to {df_clean["Vsh_Tertiary"].max():.4f}')

# ============================================================
# METHOD 3: Larionov — Pre-Tertiary (older rocks)
# Formula: Vsh = 0.33 * (2^(2 * IGR) - 1)
# Kansas rocks are Pennsylvanian age = Pre-Tertiary
# ============================================================
df_clean['Vsh_PreTertiary'] = 0.33 * (2 ** (2 * df_clean['IGR']) - 1)
df_clean['Vsh_PreTertiary'] = df_clean['Vsh_PreTertiary'].clip(0, 1)

print('\n✅ Method 3: Larionov Pre-Tertiary calculated')
print(f'   Vsh_PreTertiary range: {df_clean["Vsh_PreTertiary"].min():.4f} to {df_clean["Vsh_PreTertiary"].max():.4f}')

print('\n📌 Domain Note:')
print('   Kansas wells (like WEEKS #1-10) are typically Pennsylvanian age')
print('   So Larionov Pre-Tertiary is the most geologically appropriate!')
```

```python
# Compare the three Vshale methods
fig, axes = plt.subplots(1, 3, figsize=(14, 10), sharey=True)
fig.suptitle('📊 Vshale Comparison — Three Methods\nWEEKS #1-10, Kansas', 
             fontsize=13, fontweight='bold')

methods = [
    ('IGR',             'Linear IGR',              'steelblue'),
    ('Vsh_Tertiary',    'Larionov Tertiary',        'darkorange'),
    ('Vsh_PreTertiary', 'Larionov Pre-Tertiary ✅', 'darkgreen'),
]

for ax, (col, title, color) in zip(axes, methods):
    ax.plot(df_clean[col], df_clean['DEPTH'], color=color, linewidth=0.8)
    ax.fill_betweenx(df_clean['DEPTH'], df_clean[col], 0.5,
                     where=(df_clean[col] < 0.5),
                     color='yellow', alpha=0.3)
    ax.fill_betweenx(df_clean['DEPTH'], df_clean[col], 0.5,
                     where=(df_clean[col] >= 0.5),
                     color='brown', alpha=0.2)
    ax.axvline(0.5, color='red', linestyle='--', linewidth=1, label='Vsh = 0.5 cutoff')
    ax.set_xlim(0, 1)
    ax.invert_yaxis()
    ax.set_xlabel('Vshale', fontsize=10)
    ax.set_title(title, fontsize=10, fontweight='bold')
    ax.legend(fontsize=7)
    ax.grid(True, alpha=0.3)

axes[0].set_ylabel('Depth (ft)', fontsize=11)

plt.tight_layout()
plt.savefig('vshale_comparison.png', dpi=150, bbox_inches='tight')
plt.show()

print('📌 Key Observation:')
print('   Linear IGR gives the HIGHEST Vshale (most pessimistic)')
print('   Larionov corrections reduce Vshale (more realistic)')
print('   Pre-Tertiary correction gives the LOWEST Vshale (most optimistic for reservoirs)')
```

```python
# Scatter comparison of the three methods
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
fig.suptitle('Vshale Method Comparison — Scatter Plots', fontsize=12, fontweight='bold')

axes[0].scatter(df_clean['IGR'], df_clean['Vsh_Tertiary'], 
                alpha=0.3, s=5, color='darkorange')
axes[0].plot([0,1],[0,1],'k--', label='1:1 line')
axes[0].set_xlabel('Linear IGR')
axes[0].set_ylabel('Larionov Tertiary')
axes[0].set_title('IGR vs Larionov Tertiary')
axes[0].legend()

axes[1].scatter(df_clean['IGR'], df_clean['Vsh_PreTertiary'], 
                alpha=0.3, s=5, color='darkgreen')
axes[1].plot([0,1],[0,1],'k--', label='1:1 line')
axes[1].set_xlabel('Linear IGR')
axes[1].set_ylabel('Larionov Pre-Tertiary')
axes[1].set_title('IGR vs Larionov Pre-Tertiary ✅')
axes[1].legend()

plt.tight_layout()
plt.savefig('vshale_scatter_comparison.png', dpi=150, bbox_inches='tight')
plt.show()

print('📌 Points BELOW the 1:1 line = Larionov gives lower Vshale than linear')
print('   This is the correction effect — Larionov is more reservoir-friendly!')
```

## 🏷️ Section 5 — Lithology Classification Using Vshale Thresholds

```python
# We'll use the Pre-Tertiary Vshale for classification
# as it's most appropriate for Kansas geology

def classify_lithology(vshale):
    """
    Classify lithology based on Vshale thresholds.
    These are standard industry cutoffs used by petrophysicists.
    """
    if vshale < 0.20:
        return 'Clean Sand'      # ✅ Best reservoir quality
    elif vshale < 0.40:
        return 'Silty Sand'      # 🟡 Good reservoir quality
    elif vshale < 0.60:
        return 'Sandy Shale'     # 🟠 Poor reservoir quality
    else:
        return 'Shale'           # ❌ Non-reservoir

# Apply classification
df_clean['Lithology'] = df_clean['Vsh_PreTertiary'].apply(classify_lithology)

# Count per class
lith_counts = df_clean['Lithology'].value_counts()
lith_pct    = df_clean['Lithology'].value_counts(normalize=True) * 100

print('=== 🏷️ LITHOLOGY CLASSIFICATION RESULTS ===')
print(f'\n{"Lithology":<15} {"Count":>8} {"Percentage":>12}')
print('-' * 38)
for lith in ['Clean Sand', 'Silty Sand', 'Sandy Shale', 'Shale']:
    if lith in lith_counts:
        print(f'{lith:<15} {lith_counts[lith]:>8} {lith_pct[lith]:>10.1f}%')

print(f'\n📌 Interpretation:')
print(f'   Clean Sand + Silty Sand = potential reservoir zones')
print(f'   Sandy Shale + Shale = non-reservoir / seal zones')
```

```python
# Create a numeric label for ML (Week 3 will use this!)
lith_map = {'Clean Sand': 0, 'Silty Sand': 1, 'Sandy Shale': 2, 'Shale': 3}
df_clean['Lith_Label'] = df_clean['Lithology'].map(lith_map)

print('=== 🔢 NUMERIC LABELS (For Week 3 ML Classifier) ===')
print('   0 = Clean Sand')
print('   1 = Silty Sand')
print('   2 = Sandy Shale')
print('   3 = Shale')
print(f'\nLabel distribution:')
print(df_clean['Lith_Label'].value_counts().sort_index())
print(f'\n✅ These labels will be our TARGET variable in Week 3!')
print(f'   We will train a Random Forest / XGBoost to predict')
print(f'   these classes using GR, RHOB, DPOR, CNPOR, RILD as features.')
```
